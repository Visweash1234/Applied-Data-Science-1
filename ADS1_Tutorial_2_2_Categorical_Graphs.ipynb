```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Let's start by randomly generated 4 sets of sample data sets
# Explore what is np.random.normal ( )?
sample1 = np.random.normal(-1.0, 1.0, 10000)
sample2 = np.random.normal(1.0, 0.5, 10000)
sample3 = np.random.normal(0.0, 1.5, 10000)
sample4 = np.random.normal(-0.2, 2.0, 10000)
# Test by printing out each sample data sets, Example print(sample1)
# Find out the data type of the samples
```

```python
def plot_subplotted_histograms(sample1, sample2, sample3, sample4):
    """
    The following scripts will Plots 4 histograms as subplots. Understand what is subplots?
    """
    fig, axs = plt.subplots(2, 2, dpi=144)
    axs = axs.flatten()

    axs[0].hist(sample1)
    axs[1].hist(sample2)
    axs[2].hist(sample3)
    axs[3].hist(sample4)

    for i, ax in enumerate(axs):
        ax.set_ylabel('N')
        ax.set_title('Sample ' + str(i + 1))

    fig.subplots_adjust(wspace=0.5, hspace=0.5)
    plt.show()
    return
```

```python
plot_subplotted_histograms(sample1, sample2, sample3, sample4)
```

```python
def plot_overplotted_histograms(*samples):
    """
    Plots 4 histograms on top of each other
    """
    plt.figure(dpi=144)

    for i, sample in enumerate(samples):
        plt.hist(sample, label='Sample ' + str(i + 1), range=(-5, 5), bins=10, alpha=0.5)

    plt.ylabel('N')
    plt.xlim(-5, 5)
    plt.legend()
    plt.show()
    return
```

```python
plot_overplotted_histograms(sample1, sample2, sample3, sample4)
```

```python
df_tesco = pd.read_csv('/content/TSCO_ann.csv',index_col='year')
df_bp = pd.read_csv('/content/BP_ann.csv',index_col='year')
df_barclays = pd.read_csv('/content/BCS_ann.csv',index_col='year')
df_vodaphone= pd.read_csv('/content/VOD_ann.csv',index_col='year')
print(df_tesco.head())
print(df_bp.head())
print(df_barclays.head())
print(df_vodaphone.head())
```

```python
def plot_subplotted_annual_returns():
    """
    Plots 4 histograms as subplots for annual returns of different companies.
    """
    fig, axs = plt.subplots(2, 2, figsize=(12, 10), dpi=144)
    axs = axs.flatten()

    dataframes = [df_tesco, df_bp, df_barclays, df_vodaphone]
    titles = ['Tesco Annual Returns', 'BP Annual Returns', 'Barclays Annual Returns', 'Vodafone Annual Returns']

    for i, df in enumerate(dataframes):
        axs[i].hist(df['ann_return'], bins=20, alpha=0.7, color='skyblue', edgecolor='black')
        axs[i].set_title(titles[i])
        axs[i].set_xlabel('Annual Return (%)')
        axs[i].set_ylabel('Frequency')
        axs[i].grid(axis='y', alpha=0.75)

    plt.tight_layout()
    plt.show()
    return
```

```python
plot_subplotted_annual_returns()
```

```python
def plot_overplotted_annual_returns():
    """
    Plots 4 histograms on top of each other for annual returns of different companies.
    """
    plt.figure(figsize=(10, 7), dpi=144)

    dataframes = [df_tesco, df_bp, df_barclays, df_vodaphone]
    labels = ['Tesco', 'BP', 'Barclays', 'Vodafone']
    colors = ['blue', 'green', 'red', 'purple']

    for i, df in enumerate(dataframes):
        plt.hist(df['ann_return'], bins=20, alpha=0.5, label=labels[i], color=colors[i])

    plt.title('Overplotted Annual Returns for Various Companies')
    plt.xlabel('Annual Return (%)')
    plt.ylabel('Frequency')
    plt.legend()
    plt.grid(axis='y', alpha=0.75)
    plt.show()
    return
```

```python
plot_overplotted_annual_returns()
```

```python
def plot_gdp_pie(gdp, labels=countries):
    """
    Creates a pie chart of the GDPs of the 4 largest countries, by GDP
    """
    plt.figure(dpi=144)

    plt.pie(gdp, labels=labels, autopct='%1.1f%%')
    plt.title('Largest Economies in the EU')
    plt.axis('Equal')
    plt.show()
    return
```

```python
def plot_market_cap_pie(cap, labels):
    """
    Creates a pie chart of the market cap of 4 major companies
    """
    plt.figure(dpi=144)

    plt.pie(cap, labels=labels, autopct='%1.1f%%', startangle=90)
    plt.title('Market Capitalization of Major Companies')
    plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
    plt.show()
    return
```

```python
# Calculate market capitalization (using last price as a proxy)
cap_values = np.array([
    df_tesco['price'].iloc[-1],
    df_bp['price'].iloc[-1],
    df_barclays['price'].iloc[-1],
    df_vodaphone['price'].iloc[-1]
])

# Define labels for the companies
company_labels = ['Tesco', 'BP', 'Barclays', 'Vodafone']

# Call the function with the calculated cap values and labels
plot_market_cap_pie(cap_values, company_labels)
```

```python
def plot_market_cap_pie_unnormalised(norm_cap, labels):
    """
    Creates a pie chart of the market cap of 4 major companies, normalized to the FTSE
    """
    plt.figure(dpi=144)

    plt.pie(norm_cap, labels=labels, autopct='%1.1f%%', startangle=90, normalize=False)
    plt.title('Market Capitalization of Major Companies (Normalized to FTSE)')
    plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
    plt.show()
    return
```

```python
# Calculate the sum of the market caps to represent the 'FTSE' for normalization
ftse_total = np.sum(cap_values)

# Normalize the individual company market caps
norm_cap_values = cap_values / ftse_total

# Call the function with the normalized cap values and labels
plot_market_cap_pie_unnormalised(norm_cap_values, company_labels)
```
