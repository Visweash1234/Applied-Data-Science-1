```python
#1) Manually (inbuilt python methods)
with open('countries_top10.csv', 'r') as f:
    data = []
    for line in f.readlines():
        linelist = line.rstrip('\n').split(',')
        for i, value in enumerate(linelist):
            try:
                value = float(value)
                if value.is_integer():
                    value = int(value)
            except ValueError:
                continue
            linelist[i] = value
        data.append(linelist)
print(data)
print(type(data))
print(type(data[0]))
print(type(data[1][1]))
```

```python
# CSV method (standard library)
import csv
with open('countries_top10.csv', 'r') as f:
    csvreader = csv.reader(f)
    data = []
    for row in csvreader:
        for i, value in enumerate(row):
            try:
                value = float(value)
                if value.is_integer():
                    value = int(value)
            except ValueError:
                continue
            row[i] = value
        data.append(row)
print(data)
print(type(data))
print(type(data[0]))
print(type(data[1][1]))
```
